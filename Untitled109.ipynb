```python
# =========================
# 1. Import bibliotek
# =========================
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import roc_auc_score, confusion_matrix

sns.set(style="whitegrid")
# =========================
# 2. Wczytanie danych
# =========================
df = pd.read_excel(
    "default_of_credit_card_clients.xls",
    header=1,
    engine="xlrd"
)

# Zmiana nazwy zmiennej docelowej
df.rename(columns={
    "default payment next month": "default_payment"
}, inplace=True)
# =========================
# 3. Eksploracyjna analiza danych (EDA)
# =========================

print("Wymiary danych:", df.shape)

print("\nInformacje o danych:")
print(df.info())

print("\nBraki danych:")
print(df.isnull().sum())

print("\nStatystyki opisowe:")
print(df.describe())

print("\nRozkład zmiennej docelowej:")
print(df["default_payment"].value_counts(normalize=True))
# =========================
# 4. Wizualizacja zmiennej docelowej
# =========================

plt.figure(figsize=(5,4))
sns.countplot(data=df, x="default_payment")
plt.title("Rozkład zmiennej default_payment")
plt.show()
# =========================
# 5. Wizualizacja wybranych zmiennych
# =========================

key_features = [
    "LIMIT_BAL",
    "PAY_0",
    "PAY_2",
    "PAY_3",
    "AGE"
]

for col in key_features:
    plt.figure(figsize=(6,4))
    sns.boxplot(x="default_payment", y=col, data=df)
    plt.title(f"{col} vs default_payment")
    plt.show()
# =========================
# 6. Przygotowanie danych
# =========================

y = df["default_payment"]
X = df.drop(columns=["default_payment", "ID"])
# =========================
# 7. Podział danych
# =========================

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.30,
    random_state=42,
    stratify=y
)
# =========================
# 8. Standaryzacja
# =========================

scaler = StandardScaler()

X_train = scaler.fit_transform(X_train)
X_test = scaler.transform(X_test)
# =========================
# 9. Budowa modelu
# =========================

model = LogisticRegression(
    max_iter=1000,
    class_weight="balanced",
    random_state=42
)

model.fit(X_train, y_train)
# =========================
# 10. Predykcja i ocena
# =========================

y_prob = model.predict_proba(X_test)[:, 1]
y_pred = (y_prob >= 0.30).astype(int)

print("AUC:", roc_auc_score(y_test, y_prob))

print("\nConfusion Matrix:")
print(confusion_matrix(y_test, y_pred))
```
